```python
# sets - work on unique values
person_1_logged_in = {2,3,2,4,1,34,5,1,4,5,9,7,5}
print(person_1_logged_in) # Print the unique values of person 1
person_2_logged_in = {1,3,5,9,1,4,1,2,6,4}
print(person_2_logged_in) # Print the unique values of person 2
print(person_1_logged_in | person_2_logged_in) #print unique values from both sets
print(person_1_logged_in & person_2_logged_in) #print both sets
print(person_2_logged_in - person_1_logged_in) # Print the values which are present in person 2 but not in person 1
print(person_1_logged_in ^ person_2_logged_in) # Only print those values which are unique in both sets
```

```python
for key, value in dict_example.items():
    print(f"Key is {key} and value is {value}")
```

Nested For Loops

```python
name = ["Ashiq", "Adnan", "Rohan"]
fav_food = ["Beef", "Chicken"]
for n in name:
    for f in fav_food:
        print(f"{n} loves to eat {f}")
```

While Loops

```python
number = 0
while number < 5:
    number = number+1
    if number==3:
        break
    print(number)
    
```

```python
number = 0
while number < 5:
    number = number+1
    if number==3:
        continue
    print(number)
```

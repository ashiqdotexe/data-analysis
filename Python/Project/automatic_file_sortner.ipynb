```python
import os, shutil
path = r"E:/Practice Project/data science/data-analysis/Python/Project/"
```

```python
files = os.listdir(path=path)
files
```

```python
extension_map = {
    ".csv": "csv",
    ".jpg": "image",
    ".jfif": "image"
}
```

```python
for folder in set(extension_map.values()):
    os.makedirs(os.path.join(path, folder), exist_ok = True)
```

```python
for file in files:
    for ext, folder in extension_map.items():
        if file.lower().endswith(ext):
            src = os.path.join(path, file)
            dst = os.path.join(path, folder, file)
            shutil.move(src, dst)
            break
```

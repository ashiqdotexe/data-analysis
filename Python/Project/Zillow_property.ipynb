```python
import requests
import pandas as pd
import time
from typing import List, Dict
import json

# Top 20 US states by population with their Zillow region IDs
STATES = {
    'CA': {'name': 'California', 'regionId': 9},
    'TX': {'name': 'Texas', 'regionId': 54},
    'FL': {'name': 'Florida', 'regionId': 10},
    'NY': {'name': 'New York', 'regionId': 43},
    'PA': {'name': 'Pennsylvania', 'regionId': 45},
    'IL': {'name': 'Illinois', 'regionId': 15},
    'OH': {'name': 'Ohio', 'regionId': 44},
    'GA': {'name': 'Georgia', 'regionId': 11},
    'NC': {'name': 'North Carolina', 'regionId': 42},
    'MI': {'name': 'Michigan', 'regionId': 33},
    'NJ': {'name': 'New Jersey', 'regionId': 40},
    'VA': {'name': 'Virginia', 'regionId': 58},
    'WA': {'name': 'Washington', 'regionId': 59},
    'AZ': {'name': 'Arizona', 'regionId': 4},
    'MA': {'name': 'Massachusetts', 'regionId': 32},
    'TN': {'name': 'Tennessee', 'regionId': 53},
    'IN': {'name': 'Indiana', 'regionId': 16},
    'MO': {'name': 'Missouri', 'regionId': 35},
    'MD': {'name': 'Maryland', 'regionId': 31},
    'WI': {'name': 'Wisconsin', 'regionId': 61}
}

class ZillowScraper:
    def __init__(self):
        self.base_url = 'https://www.zillow.com/async-create-search-page-state'
        self.headers = {
            'accept': '*/*',
            'accept-language': 'en-US,en;q=0.9',
            'content-type': 'application/json',
            'origin': 'https://www.zillow.com',
            'user-agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/144.0.0.0 Safari/537.36',
        }
        self.all_listings = []
    
    def create_payload(self, state_code: str, region_id: int, page: int = 1) -> Dict:
        """Create the request payload for a given state and page"""
        return {
            'searchQueryState': {
                'isMapVisible': True,
                'mapBounds': {
                    'north': 50.0,
                    'south': 25.0,
                    'east': -65.0,
                    'west': -125.0,
                },
                'filterState': {
                    'sortSelection': {
                        'value': 'globalrelevanceex',
                    },
                },
                'isListVisible': True,
                'mapZoom': 6,
                'usersSearchTerm': state_code,
                'regionSelection': [
                    {
                        'regionId': region_id,
                        'regionType': 2,
                    },
                ],
                'pagination': {
                    'currentPage': page,
                },
            },
            'wants': {
                'cat1': ['listResults', 'mapResults'],
                'cat2': ['total'],
            },
            'requestId': page,
            'isDebugRequest': False,
        }
    
    def parse_listing(self, listing: Dict, state_code: str, state_name: str) -> Dict:
        """Extract relevant data from a listing"""
        try:
            return {
                'zpid': listing.get('zpid', ''),
                'state_code': state_code,
                'state_name': state_name,
                'address': listing.get('address', ''),
                'street': listing.get('addressStreet', ''),
                'city': listing.get('addressCity', ''),
                'zipcode': listing.get('addressZipcode', ''),
                'price': listing.get('unformattedPrice', ''),
                'price_formatted': listing.get('price', ''),
                'beds': listing.get('beds', ''),
                'baths': listing.get('baths', ''),
                'area_sqft': listing.get('area', ''),
                'latitude': listing.get('latLong', {}).get('latitude', ''),
                'longitude': listing.get('latLong', {}).get('longitude', ''),
                'status': listing.get('statusText', ''),
                'home_type': listing.get('hdpData', {}).get('homeInfo', {}).get('homeType', ''),
                'days_on_zillow': listing.get('hdpData', {}).get('homeInfo', {}).get('daysOnZillow', ''),
                'zestimate': listing.get('zestimate', ''),
                'detail_url': listing.get('detailUrl', ''),
                'has_open_house': listing.get('hasOpenHouse', False),
                'open_house_date': listing.get('openHouseStartDate', ''),
                'is_featured': listing.get('isFeaturedListing', False),
            }
        except Exception as e:
            print(f"Error parsing listing: {e}")
            return None
    
    def scrape_state(self, state_code: str, max_pages: int = 5) -> List[Dict]:
        """Scrape listings for a given state"""
        region_id = STATES[state_code]['regionId']
        state_name = STATES[state_code]['name']
        state_listings = []
        
        print(f"\n{'='*60}")
        print(f"Scraping {state_name} ({state_code})")
        print(f"{'='*60}")
        
        for page in range(1, max_pages + 1):
            try:
                print(f"  Page {page}...", end=' ')
                
                payload = self.create_payload(state_code, region_id, page)
                response = requests.put(
                    self.base_url,
                    headers=self.headers,
                    json=payload,
                    timeout=30
                )
                
                if response.status_code != 200:
                    print(f"❌ Failed (Status: {response.status_code})")
                    break
                
                data = response.json()
                
                # Extract listings
                listings = data.get('cat1', {}).get('searchResults', {}).get('listResults', [])
                
                if not listings:
                    print(f"✓ No more listings")
                    break
                
                # Parse each listing
                for listing in listings:
                    parsed = self.parse_listing(listing, state_code, state_name)
                    if parsed:
                        state_listings.append(parsed)
                
                print(f"✓ {len(listings)} listings found")
                
                # Be nice to the server
                time.sleep(2)
                
            except Exception as e:
                print(f"❌ Error: {e}")
                break
        
        print(f"Total for {state_name}: {len(state_listings)} listings")
        return state_listings
    
    def scrape_all_states(self, max_pages_per_state: int = 5) -> pd.DataFrame:
        """Scrape all top 20 states"""
        print("Starting Zillow scraper for top 20 US states...")
        print(f"Max pages per state: {max_pages_per_state}")
        
        for state_code in STATES.keys():
            listings = self.scrape_state(state_code, max_pages_per_state)
            self.all_listings.extend(listings)
            
            # Save intermediate results
            if self.all_listings:
                df_temp = pd.DataFrame(self.all_listings)
                df_temp.to_csv('E:/Projects/Zillow/zillow_data_temp.csv', index=False)
                print(f"  → Saved {len(self.all_listings)} total listings so far")
        
        # Create final DataFrame
        if self.all_listings:
            df = pd.DataFrame(self.all_listings)
            return df
        else:
            return pd.DataFrame()
    
    def save_to_csv(self, df: pd.DataFrame, filename: str = 'zillow_listings.csv'):
        """Save DataFrame to CSV"""
        if not df.empty:
            df.to_csv(f'E:/Projects/Zillow/{filename}', index=False)
            print(f"\n{'='*60}")
            print(f"✓ Saved {len(df)} listings to {filename}")
            print(f"{'='*60}")
            
            # Print summary statistics
            print("\nSummary by State:")
            print(df['state_name'].value_counts())
            print(f"\nAverage Price: ${df['price'].mean():,.2f}")
            print(f"Price Range: ${df['price'].min():,.0f} - ${df['price'].max():,.0f}")
        else:
            print("No data to save!")


def main():
    # Initialize scraper
    scraper = ZillowScraper()
    
    # Scrape all states (adjust max_pages_per_state as needed)
    # WARNING: Setting this too high may get you blocked!
    df = scraper.scrape_all_states(max_pages_per_state=3)
    
    # Save to CSV
    scraper.save_to_csv(df, 'zillow_top20_states.csv')
    
    # Also save to outputs for user access
    if not df.empty:
        df.to_csv('zillow_top20_states.csv', index=False)
        print("\n✓ File also saved to outputs directory")


if __name__ == '__main__':
    main()
```
